```python
import kairo
import pandas as pd

import warnings
warnings.filterwarnings("ignore")

from dotenv import load_dotenv
from pathlib import Path
import os

load_dotenv(Path("../..") / ".env")
```

```python
log = kairo.read_log("../../data/thesis/general/running_example/L3.xes",
    case_id = "case:concept:name",
    activity = "concept:name",
    timestamp = "time:timestamp",
    #event_id = "event:uidd",
    start_timestamp = "start_timestamp",
    resource = "org:resource",
    #event_duration = "event:runtime_min")
    )
log.head()
```

```python
stats = kairo.compute_log_stats(log)
stats
```

```python
activity_plot = kairo.plot_activity_counts(stats)
activity_plot.show()
```

```python
features = kairo.compute_features_intra(log, sliding_window_size=1)
features
```

```python
std_features = kairo.standardize(features, method="minmax")
std_features
```

```python
pca = kairo.compute_pca(std_features, num_components=50)
pca_plot = kairo.plot_pca_variances(pca)
pca_plot.show()
#pca
```

```python
reconstruction_errors = kairo.compute_pca_reconstruction_errors(std_features, pca)
reconstruction_plot = kairo.plot_pca_reconstruction_errors(reconstruction_errors)
reconstruction_plot.show()
```

```python
compressed_features = kairo.apply_pca(std_features, pca, cut_component=15)
compressed_features
```

```python
pca_plot_cut = kairo.plot_pca_variances(pca, cut_component=15)
pca_plot_cut.show()
```

```python
reconstruction_plot_cut = kairo.plot_pca_reconstruction_errors(reconstruction_errors, cut_component=15)
reconstruction_plot_cut.show()
```

```python
# heuristic: min_samples * num_dimensions (not working good here)
# try around min_samples >= D + 1 where D is the number of PCA dimensions
k_distances_plot = kairo.plot_dbscan_k_distance(compressed_features, min_samples=234, distance="euclidean")
k_distances_plot.show()
```

```python
dbscan = kairo.compute_dbscan(compressed_features, min_samples=234, eps=0.7, distance="euclidean")
clusters = kairo.get_dbscan_states(compressed_features, dbscan)
clusters
```

```python
clustering_scores = [
    {"type": "DBCV", "value": kairo.compute_dbcv(compressed_features, clusters, distance=dbscan.metric)},
    {"type": "Silhouette", "value": kairo.compute_silhouette(compressed_features, clusters, distance=dbscan.metric)},
    {"type": "Calinski-Harabasz", "value": kairo.compute_calinski_harabasz(compressed_features, clusters)},
]
clustering_scores
```

```python
features_and_clusters = pd.concat([compressed_features, clusters], axis=1)
features_and_clusters
```

```python
state_distances = kairo.get_dbscan_state_distances(dbscan)
state_distances
```

```python
state_freqs = kairo.get_dbscan_state_frequencies(features_and_clusters)
state_freqs
```

```python
distances_plot = kairo.plot_dbscan_state_distances(state_distances)
distances_plot.show()
```

```python
state_colors = kairo.get_dbscan_state_colors(dbscan)
```

```python
freqs_plot = kairo.plot_dbscan_state_frequencies(state_freqs, state_colors)
freqs_plot.show()
```

```python
case_trajectory = kairo.get_dbscan_case_trajectory(features_and_clusters, case_id="case_000001")
case_trajectory
```

```python
case_trajectory_plot = kairo.plot_dbscan_case_trajectory(case_trajectory, state_colors, case_id="case_000001")
case_trajectory_plot
```

```python
case_trajectories = kairo.get_dbscan_case_trajectories(features_and_clusters, max_cases=10)
case_trajectories
```

```python
case_trajectories_plot = kairo.plot_dbscan_case_trajectories(case_trajectories, state_colors)
case_trajectories_plot
```

```python
freq_distributions = kairo.compute_state_distributions(features_and_clusters, signal_window="30D")
freq_distributions
```

```python
distributions_plot = kairo.plot_state_distributions(freq_distributions, state_colors)
distributions_plot.show()
```

```python
drift_signal = kairo.compute_drift_signal(freq_distributions)
drift_signal
```

```python
drift_signal_plot = kairo.plot_drift_signal(drift_signal)
drift_signal_plot.show()
```

```python
# the mean and the 0.9-quantile of the drift scores, the quantile as a first threshold for a drift
drift_mean = kairo.compute_drift_signal_mean(drift_signal)
drift_quantile = kairo.compute_drift_signal_quantile(drift_signal, t=0.9)
drift_mean, drift_quantile
```

```python
# the events behind every drift score, with their mean and their 0.1-quantile
signal_window_rows = kairo.compute_signal_window_rows(features_and_clusters, signal_window="30D")
rows_mean = kairo.compute_signal_window_rows_mean(signal_window_rows)
rows_quantile = kairo.compute_signal_window_rows_quantile(signal_window_rows, t=0.1)
rows_mean, rows_quantile
```

```python
signal_window_rows_plot = kairo.plot_signal_window_rows(signal_window_rows)
signal_window_rows_plot.show()
```

```python
abs_log_stats = kairo.abstract_log_stats(stats)
abs_log_stats
```

```python
abs_features = kairo.abstract_features(features)
abs_features
```

```python
abs_pca = kairo.abstract_pca(pca, cut_component=13)
abs_pca
```

```python
abs_states = kairo.abstract_states(state_freqs, state_distances, state_colors, clustering_scores=clustering_scores)
abs_states
```

```python
abs_case_trajectory = kairo.abstract_case_trajectory(case_trajectory, "case_000001")
abs_case_trajectory
```

```python
abs_case_trajectories = kairo.abstract_case_trajectories(case_trajectories)
abs_case_trajectories
```

```python
abs_distributions = kairo.abstract_state_distributions(freq_distributions)
abs_distributions
```

```python
abs_drift_signal = kairo.abstract_drift_signal(drift_signal, divergence="kl", reference="recent")
abs_drift_signal
```

```python
llm = kairo.LLMConnector(
    provider="custom",
    #api_key=os.environ["ANTHROPIC_API_KEY"],
    model="Qwen/Qwen3-VL-32B-Instruct",
    base_url="http://137.226.117.70:8000/v1"
)
```

```python
chat_history = []
total_out_tokens = 0
```

```python
# For initial request
messages = kairo.create_messages(provider="custom",
                                system_prompt=kairo.DEFAULT_SYSTEM_PROMPT, 
                                prompts=[abs_log_stats + abs_features + 
                                    abs_pca + abs_states + abs_case_trajectory + abs_case_trajectories
                                    + abs_distributions + abs_drift_signal + 
                                    "Analyze and summarize the given information. Do not use markdown format, just raw text"],
                                plots=[activity_plot, pca_plot_cut, distances_plot, freqs_plot,
                                            case_trajectory_plot, case_trajectories_plot,
                                            distributions_plot, drift_signal_plot])
chat_history += messages

num_tokens = kairo.count_input_tokens(provider="custom", messages=chat_history)
num_tokens
```

```python
# For follow up request
messages = kairo.create_messages(provider="custom",
                                system_prompt=None,
                                prompts=["what was my previous question very briefly"],
                                plots=[])
chat_history += messages

num_tokens = kairo.count_input_tokens(provider="custom", messages=chat_history)
num_tokens
```

```python
response = llm.call(messages=chat_history, max_tokens=64000)

answer = kairo.get_response_text(response)

chat_history.append({"role": "assistant", 
                     "content": answer})

print(answer)
```

```python
out_tokens = kairo.count_output_tokens("custom", response)
total_out_tokens += out_tokens
total_out_tokens
```
