```python
from openai import OpenAI

client = OpenAI(
    base_url="http://127.0.0.1:1234/v1",
    api_key="not-needed",
)
```

```python
models = client.models.list()
for model in client.models.list():
    print(model.id)
```

```python
import requests

r = requests.get("http://127.0.0.1:1234/api/v1/models")
data = r.json()
```

```python
for model in data["models"]:
    print(
        model["key"],
        "LOADED" if model["loaded_instances"] else "not loaded"
    )
```

```python
response = client.chat.completions.create(
    model="google/gemma-4-e4b",
    messages=[
        {
            "role": "user",
            "content": "Comprehensively explain me process mining."
        }
    ],
    temperature=0,
    max_tokens=1000,
)

print(response.choices[0].message.content)
```
